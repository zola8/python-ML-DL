```python
import pandas as pd

df = pd.read_csv("../../data/heart.csv")
df.head()
```

```python
X = df.drop("target", axis=1)
y = df['target']
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.4, random_state=9)
```

### Scale-Insensitive

```python
from sklearn.ensemble import RandomForestClassifier

forest = RandomForestClassifier()
forest.fit(X_train, y_train)
```

```python
from sklearn.naive_bayes import GaussianNB

nb_clf = GaussianNB()
nb_clf.fit(X_train, y_train)
```

```python
from sklearn.ensemble import GradientBoostingClassifier

gb_clf = GradientBoostingClassifier()
gb_clf.fit(X_train, y_train)
```

### Scale-Sensitive

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
from sklearn.neighbors import KNeighborsClassifier

knn = KNeighborsClassifier()
knn.fit(X_train_scaled, y_train)

# model focuses on input (we should scale the input data + test data)
# output doesn't matter; (y_train = output)
```

```python
from sklearn.linear_model import LogisticRegression

log = LogisticRegression()
log.fit(X_train_scaled, y_train)
```

```python
from sklearn.svm import SVC

svc = SVC()
svc.fit(X_train_scaled, y_train)
```

### Evaluate the default models

```python
forest.score(X_test, y_test)
```

```python
nb_clf.score(X_test, y_test)
```

```python
gb_clf.score(X_test, y_test)
```

```python
knn.score(X_test_scaled, y_test)
```

```python
log.score(X_test_scaled, y_test)
```

```python
svc.score(X_test_scaled, y_test)
```

Accuracy = 100 guess - 80 correct = 80% = 0.80 accuracy

Precision = 100 guess, 50 has disease -> let's say 20% was correct = 0.20 precision

Recall = I want to spot every patient with heart disease

So... it's better (?) to check recall, if I give you a false positive than a false negative.

```python
from sklearn.metrics import recall_score

print("--- recall scores ---")

y_preds = forest.predict(X_test)
print("Forest:               ", recall_score(y_test, y_preds))

y_preds = nb_clf.predict(X_test)
print("NB:                   ", recall_score(y_test, y_preds))

y_preds = gb_clf.predict(X_test)
print("Gradient Boost:       ", recall_score(y_test, y_preds))

y_preds = knn.predict(X_test_scaled)
print("KNN:                  ", recall_score(y_test, y_preds))

y_preds = log.predict(X_test_scaled)
print("Logistic Regression:  ", recall_score(y_test, y_preds))

y_preds = svc.predict(X_test_scaled)
print("SVC:                  ", recall_score(y_test, y_preds))
```

```python
from sklearn.metrics import roc_curve, roc_auc_score
from matplotlib import pyplot as plt

def plot_aaa(model, model_name):
    y_probs = model.predict_proba(X_test)[:, 1]
    fpr, tpr, thresholds = roc_curve(y_test, y_probs)

    plt.plot(fpr, tpr)
    plt.xlabel('False Positive Rate')
    plt.ylabel('True Positive Rate (Recall)')
    plt.title('ROC Curve - ' + model_name)
    plt.show()

    print("ROC-AUC score: ", roc_auc_score(y_test, y_probs))
```

```python
plot_aaa(forest, 'Random Forest')
```

```python
plot_aaa(log, 'Logistic Regression')
```

```python
plot_aaa(knn, 'K-Nearest Neighbors')
```

# Hyper parameter tuning

```python
from sklearn.model_selection import GridSearchCV

param_grid = {
    'n_estimators': [100, 200, 500, 600],
    # 'max_depth': [None, 10, 20, 30],
    # 'min_samples_split': [2, 5, 10],
    # 'min_samples_leaf': [1, 2, 4],
    # 'max_features': ['sqrt', 'log2', None],
}

forest2 = RandomForestClassifier(n_jobs=-1, random_state=9)

grid_search = GridSearchCV(forest2, param_grid, cv=3)
```

```python
grid_search.fit(X_train, y_train)
```

```python
grid_search.best_params_
```

```python
best_model = grid_search.best_estimator_
```

# Feature Importance Analysis

```python
feature_importances = best_model.feature_importances_
feature_names = best_model.feature_names_in_
```

```python
feature_importances, feature_names
```

```python
import numpy as np

sorted_idx = np.argsort(feature_importances)
sorted_features = feature_names[sorted_idx]
sorted_importances = feature_importances[sorted_idx]
# sorted_zip = ...

colors = plt.cm.YlGn(sorted_importances / max(sorted_importances))
```

```python
plt.barh(sorted_features, sorted_importances, color=colors)
plt.xlabel('Feature importance')
plt.ylabel('Features')
plt.show()
```
